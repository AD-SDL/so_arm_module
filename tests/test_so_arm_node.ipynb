```python
"""SO-Arm Node Test Script — run each section manually to test node actions."""

from madsci.client.node.rest_node_client import RestNodeClient
from madsci.common.types.action_types import ActionRequest
from madsci.common.types.location_types import LocationArgument
```

```python
NODE_URL = "http://localhost:2000"
POLICY_PATH = (
    "/home/vision/humanoids/lerobot_workspace/outputs"
    "/cube_bowl_training_150ep/checkpoints/last/pretrained_model"
)

client = RestNodeClient(url=NODE_URL)
```

```python
# ---------------------------------------------------------------------------
# Check node state
# ---------------------------------------------------------------------------

print(client.get_state())
```

```python
# ---------------------------------------------------------------------------
# getJ — read current joint positions
# ---------------------------------------------------------------------------

request = ActionRequest(action_name="getJ", args={})
print(client.send_action(request))
```

```python
# ---------------------------------------------------------------------------
# home — move all joints to zero
# ---------------------------------------------------------------------------

request = ActionRequest(action_name="home", args={})
print(client.send_action(request))
```

```python
# ---------------------------------------------------------------------------
# moveJ — move to joint configuration
# ---------------------------------------------------------------------------

request = ActionRequest(
    action_name="moveJ",
    args={
        "location": LocationArgument(
            location={
                "shoulder_pan.pos": 0.4835164835164835,
                "shoulder_lift.pos": -81.45054945054945,
                "elbow_flex.pos": 74.81318681318682,
                "wrist_flex.pos": 46.76923076923077,
                "wrist_roll.pos": 8.659340659340659,
                "gripper.pos": 0.0,
            }
        ).model_dump(mode="json"),
    },
)
print(client.send_action(request))
```

```python
# ---------------------------------------------------------------------------
# load_policy — load an ACT policy | Does not execute
# ---------------------------------------------------------------------------

request = ActionRequest(
    action_name="load_policy",
    args={"policy_path": POLICY_PATH},
)
print(client.send_action(request))
```

```python
# ---------------------------------------------------------------------------
# run_episode — run the already-loaded policy
# ---------------------------------------------------------------------------

request = ActionRequest(action_name="run_episode", args={})
print(client.send_action(request))

# with custom episode length
request = ActionRequest(
    action_name="run_episode",
    args={"episode_length": 20.0},
)
print(client.send_action(request))
```

```python
# ---------------------------------------------------------------------------
# pick_and_place_cube — run cube-to-bowl policy
# ---------------------------------------------------------------------------

request = ActionRequest(action_name="pick_and_place_cube", args={})
print(client.send_action(request))

# with custom episode length
request = ActionRequest(
    action_name="pick_and_place_cube",
    args={"episode_length": 20.0},
)
print(client.send_action(request))
```

```python
# ---------------------------------------------------------------------------
# run_task — run any policy by path
# ---------------------------------------------------------------------------

request = ActionRequest(
    action_name="run_task",
    args={
        "policy_path": POLICY_PATH,
        "episode_length": 15.0,
    },
)
print(client.send_action(request))
```
